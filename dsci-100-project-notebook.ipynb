#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
stat_bas <- tidy_bas |>
select (conf_rank, conf_wins_pct, total_wins)
max <- summarise (stat_bas, across(conf_rank:total_wins, ~ max(.x, na.rm = TRUE)))
min <- summarise (stat_bas, across(conf_rank:total_wins, ~ min(.x, na.rm = TRUE)))
mean <- summarise (stat_bas, across(conf_rank:total_wins, ~ mean(.x, na.rm = TRUE)))
bind <- bind_rows ( max |> mutate (stat = "max"),
           min |> mutate (stat = "min"),
           mean |> mutate (stat = "mean"))
summary <- bind |> select (stat, conf_rank, conf_wins_pct, total_wins)
summary
```

Visualization 1: This visualization shows the relationship between conference win percentage and bid types (at-large or auto). The plot suggests that teams that received at-large bids generally had win percentages between 60% and 80%, while teams received auto bids tend to have higher win percentages, often closer to 100%. The differences in the two distributions show that conference win percentage may be a useful predictor of bid type. However, since the prediction seems to overlap, we need to consider other predictors to make more accurate prediction

```r
options (repr.plot.width = 10 , repr.plot.height = 5)
viz_1 <- tidy_bas |>
ggplot (aes (x = conf_wins_pct, fill = bid)) +
geom_histogram (position = "dodge", bins = 20) +
labs (x= "Conference Wins Percentage", y = "Count", 
      title = "Distribution of conference win percentage by Bid Type")+
theme (axis.title.x = element_text(size = 17),
       axis.title.y = element_text(size = 17), 
       plot.title = element_text (size = 17)) +
facet_grid (~bid)
viz_1
```

Visualization 2: This stacked bar plot compares the percentage of teams receiving each type of bid at different conference rankings (rank 1st to 10th). Higher-ranked teams tend to receive more auto bids, while lower-ranked teams tend to receive more at-large bids. This is relevant to the analysis as it helps me know that the higher the team's rank, the more likely it will receive an auto bid. Therefore, conference rank can be a useful predictor to predict what kind of bid a team will receive.

```r
options (repr.plot.width = 6 , repr.plot.height = 5)
viz_2 <- tidy_bas |>
mutate (conf_rank = as_factor (conf_rank),)|>
filter (!is.na (conf_rank))|>
ggplot (aes (y= conf_rank, fill = bid)) +
geom_bar (position = "fill")+
labs (title = "Relationship between Conference ranking and Bid types",
      y = "Conference Ranking",
      x = "Proportion",
     fill = "Bid Type")+
theme (axis.title.x = element_text(size = 17),
       axis.title.y = element_text(size = 17), 
       plot.title = element_text (size = 13)) 

viz_2
```
